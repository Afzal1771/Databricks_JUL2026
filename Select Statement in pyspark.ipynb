####Select()

Select() is used to select single or multiple column from dataframe

####Create emp_df dataframe from emp delta_table

```python
emp_df = spark.sql("select * from workspace.test_schema.employee")
     
```

####Display data from dataframe

```python
emp_df.show()
```

####display few column from emp_df

```python
emp_df.select("EMPNO","ENAME","SAL").show()
```

####Display all columns from dataframe

```python
emp_df.alias("emp").select("emp.*").show()
```

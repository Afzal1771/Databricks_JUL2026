#Broadcast


Broadcast is optimization technique that send small Dataframe to all workers node on cluster
instead of shuffling large amount of dataframe on cluster.

spark keeps copies of small dataset on each executors.

**Where to use broadcast?**

1. one table is small and table is large then small table fit in executor memory.

**why use it ?**

1. it's improve query performance.
2. it's reduce execution time and reduce cluster resources,

####path declaration

```python
file_1 = "/Volumes/workspace/boradcast_schema/broadcasting_files/dept_2.csv"

file_2 = "/Volumes/workspace/boradcast_schema/broadcasting_files/emp_2.csv"
```

####create emp_df Dataframe

```python
emp_df = spark.read.csv(file_2, header=True, inferSchema=True)
display(emp_df)

```

####create dept Dataframe

```python
dept_df = spark.read.csv(file_1, header=True, inferSchema=True)
display(dept_df)
```

####Braodcast optimization -

```python
from pyspark.sql.functions import broadcast
```

####Display data with broadcasting

```python
emp_dept_df_broadcast = emp_df.join(broadcast(dept_df), "deptno","inner")
display(emp_dept_df_broadcast)
```


####Display data without broadcasting

```python
emp_dept_df_without_broadcast = emp_df.join(dept_df, "deptno","inner")
display(emp_dept_df_without_broadcast )
```

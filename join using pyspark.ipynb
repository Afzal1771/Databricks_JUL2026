```python
# Employee Data
employee_data = [
    (101, "Alice", 25, 50000, 1),
    (102, "Bob", 30, 60000, 2),
    (103, "Charlie", 28, 55000, 1),
    (104, "David", 35, 70000, 3),
    (105, "Eva", 27, 65000, 2)
]

employee_columns = ["EmpID", "EmpName", "Age", "Salary", "DeptID"]

# Create Employee DataFrame
employee_df = spark.createDataFrame(employee_data, employee_columns)

# Department Data
dept_data = [
    (1, "HR"),
    (2, "IT"),
    (3, "Finance")
]

dept_columns = ["DeptID", "DeptName"]

# Create Department DataFrame
dept_df = spark.createDataFrame(dept_data, dept_columns)

# Show DataFrames
print("Employee DataFrame")
employee_df.show()

print("Department DataFrame")
dept_df.show()
```

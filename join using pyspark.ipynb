```python
# Employee Data
employee_data = [
    (101, "Alice", 25, 50000, 1),
    (102, "Bob", 30, 60000, 2),
    (103, "Charlie", 28, 55000, 1),
    (104, "David", 35, 70000, 3),
    (105, "Eva", 27,665000, 2),
    (106, "Tony", 29, 65000, 7),
    (107, "clark", 37, 65000, 8),
    (108, "Jammy", 21, 65000, 5)
]

employee_columns = ["EmpID", "EmpName", "Age", "Salary", "DeptID"]

# Create Employee DataFrame
employee_df = spark.createDataFrame(employee_data, employee_columns)

# Department Data
dept_data = [
    (1, "HR"),
    (2, "IT"),
    (3, "Finance"),
    (4, "Marketing"),
    (5, "Sales")
]

dept_columns = ["DeptID", "DeptName"]

# Create Department DataFrame
dept_df = spark.createDataFrame(dept_data, dept_columns)

# Show DataFrames
print("Employee DataFrame")
employee_df.show()

print("Department DataFrame")
dept_df.show()
```

Types - 
1. Inner join
2. left join
3. right join
4. full join
5. left semi join
6. left anti join

```python
#Inner Join - inner join pulled only matching records from both table.

employee_df.alias("emp").join(dept_df.alias("dept"),"DeptID", "inner").show()



```

```python
#Left join - left join pulled all records from left table and only matching records from right table and also display null value insted of non- matching row 

employee_df.alias("emp").join(dept_df.alias("dept"),"DeptID", "left_outer").show()
```

```python
#Right join - right join pulled all records from right table and only matching records from left table and also display null value instead of non- matching row in left side

employee_df.alias("emp").join(dept_df.alias("dept"),"DeptID","Right").display()
```

```python
#Full join - full join pulled all records from both table and also display null value instead of non- matching row in both side

employee_df.alias("emp").join(dept_df.alias("dept"),"DeptID","Full").display()
```

```python
#Semi Jon - Semi Join pulled records from left table that records are matching with right table.

employee_df.alias("emp").join(dept_df.alias("dept"),"DeptID","Semi").display()
```

```python
#Anti Jon - Semi Join pulled records from right table that records are not matching with left table.

employee_df.alias("emp").join(dept_df.alias("dept"),"DeptID","Anti").display()
```

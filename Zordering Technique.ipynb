#Z-ordering

1. z-ordering is optimization technique in databricks for delta table.
2. it's reorganized data based on one or more column, improving data skipping , reducing data scanning and 
improve query performance.


1. Zordering helps co_locate related data in the same file.
2. when you are frequently filter or join on column, zordering arrange the data so those value are stored closer on the 
disk. fewer files need to read.

```python
df = spark.sql("select * from workspace.boradcast_schema.emp")
```

```python
df.display()
```

```python
df.write.format("delta").mode("overwrite").save("/Volumes/workspace/boradcast_schema/broadcasting_files/zordering/")
```

```python
spark.sql("""OPTIMIZE delta.`/Volumes/workspace/boradcast_schema/broadcasting_files/zordering/`
ZORDER BY (deptno)
""")
```

```python
%sql
optimize workspace.boradcast_schema.emp
zorder by (deptno)
```

#Repartition

1. Repartition in Databricks is used to increase or decrease no of partition in Dataframe.
2. It's perform full shuffling of data to distribute it to same partition.<br>

**Note -** A full shuffle in repartition() means that Spark redistributes data across all executors to create new partitions. During this process, rows are moved over the network so they can be evenly distributed or grouped by a specified column.

####When to use repartition()

1. Before writing data to control the number of output files.
2. Before joins or aggregations to improve data distribution.
3. When you need to increase the number of partitions for better parallelism.

#Create DataFrame

```python
df = spark.read.format("csv")\
    .option("header","true")\
    .option("inferSchema","true")\
    .load("/Volumes/workspace/test_schema/volumes/Emp.csv")
```

#### Display data from df dataframe

```python
display(df)
```

Create delta table using DF dataframe

```python
df.repartition(3).write.format("delta")\
    .option("header","true")\
    .mode("overwrite")\
        .saveAsTable("workspace.test_schema.employee")
   
```

Display data using Delta table

```python
%sql
select * from workspace.test_schema.employee
```

Create Delta Lake Using DF DataFrame 

```python
df.repartition(3).write.format("delta")\
    .option("header","true")\
    .mode("overwrite")\
        .save("/Volumes/workspace/test_schema/volumes/employee/2026_08/")
```

####Create DataFrame

```python
employee_data = [
    (101, "Alice", "HR", 50000),
    (102, "Bob", "IT", 65000),
    (103, "Charlie", "HR", 55000),
    (104, "David", "Finance", 70000),
    (105, "Eva", "IT", 65000),
    (106, "Frank", "IT", 75000)
]

columns = ["EmpID", "EmpName", "DeptName", "Salary"]

employee_df = spark.createDataFrame(employee_data, columns)
employee_df.show()
```

####Windows Row_nubmer()

In Row_num() if value are same then assign diffrent rank 

```python
from pyspark.sql.functions import row_number , rank , dense_rank
from pyspark.sql.window import Window
```

```python
rownum_specification = Window.orderBy("Salary")
Rank_df = employee_df.withColumn("RN", row_number().over(rank_specification))

display(Rank_df)
```

####Windows Rank()

if value are same then assign same rank and also skip next rank

```python
rank_specification = Window.orderBy("Salary")

Rank_df = employee_df.withColumn("Rn" , rank().over(rank_specification))

display(Rank_df)
```


####Dense_Rank()

if value are same then assign same rank and also don't skip next from position

```python
Dense_rank_specification = Window.orderBy("Salary")

Dense_rank = employee_df.withColumn("RN",dense_rank().over(Dense_rank_specification))

display(Dense_rank)
```

```python
emp_df = spark.sql("select * from workspace.test_schema.employee")
```

Write a programe to find highest salary from emp_df

```python
from pyspark.sql.functions import dense_rank , desc , asc
from pyspark.sql.window import Window

high_Salary = Window.orderBy(desc("Sal"))

High_sal_DF = emp_df.withColumn("RN",dense_rank().over(high_Salary))\
    .filter("RN = 1")

display(High_sal_DF)
```

Write a programe to find highest salary in each deptno

```python
from pyspark.sql.functions import dense_rank , desc
from pyspark.sql.window import Window

high_Salary_each_group = Window.partitionBy("deptno").orderBy(desc("Sal"))

High_sal_DF_each_group = emp_df.withColumn("RN",dense_rank().over(high_Salary_each_group))\
    .filter("RN = 1")

display(High_sal_DF_each_group)
```


Write a programe to find top 3 salary from emp_df

```python
from pyspark.sql.functions import dense_rank , desc
from pyspark.sql.window import Window

Top3_Salary_each_group = Window.orderBy(desc("Sal"))

Top_3_salary = emp_df.withColumn("RN",dense_rank().over(Top3_Salary_each_group))\
    .filter("RN <= 3")

display(Top_3_salary)
```

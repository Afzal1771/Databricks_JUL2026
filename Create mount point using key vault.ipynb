
####Some step for creation mount point using key vault

1. create databricks workspace
2. create cluster within Databricks workspace.
3. create storage account name = newexl and his container.
4. create test-app by using registration and gran access by storage account.
5. create key-vault , create client id and secret id within key vault.
6. grant RBAC key vault admin access to user and key vault secret user access to AzureDatabricks
7, create scope.

#unmount point

```python
dbutils.fs.unmount("/mnt/bronze")
```

####Mount point creation

```python
configs = {
    "fs.azure.account.auth.type": "OAuth",
    "fs.azure.account.oauth.provider.type": "org.apache.hadoop.fs.azurebfs.oauth2.ClientCredsTokenProvider",
    "fs.azure.account.oauth2.client.id": dbutils.secrets.get(scope="scopeconnection", key="client-id"),
    "fs.azure.account.oauth2.client.secret": dbutils.secrets.get(scope="scopeconnection", key="secret-id"),
    "fs.azure.account.oauth2.client.endpoint": "https://login.microsoftonline.com/65583477-d432-469c-b4af-ab046853916f/oauth2/token","createRemoteFileSystemDuringIinitialization": "true"
}

dbutils.fs.mount(
    source="abfss://bronze@newexl.dfs.core.windows.net/",
    mount_point="/mnt/bronze",
    extra_configs=configs
)
```

####Check file name by using databricks utilities

```python
dbutils.fs.ls("/mnt/bronze")
```

##Partiton 

1. Partition is techinique in databricks that physically organize data into small file based on the value. <br>
2. Partition allows spark to read only required data instead of scanning all data from table. <br>

**Advantage and Disadvantage of partition**<br>

**Advantage**<br>
1. Parition improved query performance.<br>
2. Partition reduced query execution time.<br>
3. Partition provide less data scanned and scanned only required data.

**Disdvantage**
1. Too many Partition creates many small files.<br>
2. increases metadata overhead.<br>
3. Changing partition strategy usually required to recreate table.

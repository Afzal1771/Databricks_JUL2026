##Partiton 

1. Partition is techinique in databricks that physically organize data into small file based on the value. <br>
2. Partition allows spark to read only required data instead of scanning all data from table. <br>

**Advantage and Disadvantage of partition**<br>

**Advantage**<br>
1. Parition improved query performance.<br>
2. Partition reduced query execution time.<br>
3. Partition provide less data scanned and scanned only required data.

**Disdvantage**
1. Too many Partition creates many small files.<br>
2. increases metadata overhead.<br>
3. Changing partition strategy usually required to recreate table.


### Create dataframe for Employee File 

```python
df = spark.read.format("csv")\
    .option("header", "true")\
    .option("inferSchema", "true")\
        .load("/Volumes/workspace/test_schema/volumes/Emp.csv")
```

###Display data from df dataframe

####Pull data from DF dataframe where deptno = 10

```python
filter_data_10 =  df.filter(df.DEPTNO=="10")\
    .select("EMPNO","ENAME","SAL","DEPTNO")

display(filter_data_10)

    
```

####Pull Data from DF DataFrame for deptno 20 and 30

```python
filter_data_20_30 = df.filter(df.DEPTNO.isin("30","20"))\
    .select("EMPNO","ENAME","SAL","DEPTNO")

display(filter_data_20_30)
```

**Note :**  Create emp_csv file and load data based on deptno partition in csv format 

```python
df.write.format("csv")\
    .option("header","true")\
        .partitionBy("DEPTNO")\
        .mode("overwrite")\
            .save("/Volumes/workspace/second_schema/volumes/emp_csv")
```
